# Atividade I
Escreva um programa em Python que gere um número aleatório de 0 a 15, usando a biblioteca 'random' e o método 'randint()'. Na sequência, utilizando um loop faça com que o usuário tente acertar o número gerado.
A cada vez que o número é acertado, armazene num arquivo o número gerado e a quantidade de tentativas que o usuário realizou para acertar. Quando o usuário decidir parar de adivinhar, mostre todo o histórico que ele construiu durante a execução do programa.

```python
import random
import csv

great_break = "==================="
small_break = "-------------------"


with open('./sample_data/jogatina.csv', 'w') as file:
  writer = csv.DictWriter(file, fieldnames=['resposta', 'qtd_tentativas'])
  writer.writeheader()

  qtd_jogatinas = 0
  quit = False

  while quit is False:
    print(great_break)
    qtd_jogatinas += 1
    print(f'Iniciando jogatina #{qtd_jogatinas}...')
    print(small_break)

    resposta = random.randint(1, 15)

    qtd_tentativas = 0
    while True:
      tentativa = int(input(f'Digite a sua tentativa #{qtd_tentativas+1} (0 para sair): '))
      if tentativa == 0:
        quit = True
        break

      qtd_tentativas += 1
      if tentativa == resposta:
        print(f'Boa! Você acertou em {qtd_tentativas} tentativas! A resposta era {resposta} mesmo!')
        print(small_break)

        writer.writerow({
            "resposta": resposta,
            "qtd_tentativas": qtd_tentativas,
          })

        break
      else:
        print("Não foi dessa vez! Tente novamente.")
        print(small_break)


  print(great_break)
  print("Calculando resultado da jogatina...")


with open('./sample_data/jogatina.csv') as file:
  reader = csv.DictReader(file)

  i = 1
  for row in reader:
    print(f'Jogatina #{i}: A resposta era {row['resposta']}. Descoberto em {row["qtd_tentativas"]} tentativas.')
    i += 1

```

# Atividade II
Crie um programa que permita adicionar itens para uma lista de compras.
O usuário vai adicionando a quantidade e o valor de cada item da lista. Ao final, deve mostrar o valor total das compras e salvar a lista num arquivo .CSV.

```python
import csv


print(great_break)

lista = []

i = 0
while True:
  i += 1
  print(small_break)

  print(f'Produto #{i}')
  qtd = int(input('Digite a quantidade de itens (0 para sair): '))
  if qtd == 0:
    break

  valor = float(input('Digite o valor unitário: '))

  lista.append({
      "cd": i,
      "qtd": qtd,
      "valor": valor
  })

  print(small_break)

total = 0
for item in lista:
  total += item['qtd'] * item['valor']

print(great_break)
print(f'Valor total: {total}')
print(great_break)

with open('./sample_data/lista_compras.csv', 'w') as file:
  writer = csv.DictWriter(file, fieldnames=['cd', 'qtd', 'valor'])
  writer.writeheader()
  writer.writerows(lista)
```
