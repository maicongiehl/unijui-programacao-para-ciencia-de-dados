```python
# 1. Buscar o dataset e 2. Inserir no pandas
import pandas as pd

nomes = [
    'Ácido Fixo', # "fixed acidity",
    'Ácido Volátil', # "volatile acidity",
    'Ácido Cítrico', # "citric acid",
    'Açúcar Residual', # "residual sugar",
    'Clorídeos', # "chlorides",
    'Dióxido de Enxofre Livre', # "free sulfur dioxide",
    'Dióxido de Enxofre Total', # "total sulfur dioxide",
    'Densidade', # "density",
    'pH', # "pH",
    'Sulfato', # "sulphates",
    'Álcool', # "alcohol",
    'Qualidade', # "quality"
]

df = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv', names = nomes, sep=';', skiprows=1)
df.head()
```

```python
# 3. Inserir uma coluna adicional com a classsificação ( de acorco com a tabela abaixo):

# Baixa: nota menor ou igual a 4.
# Média: nota 5 ou 6.
# Alta: nota maior ou igual a 7.
# Obs: Essa classificação é para efeitos academicos

import numpy as np

condicoes = [df["Qualidade"] >= 7, df["Qualidade"] >= 5, df["Qualidade"] < 5]
resultados = ["Alta", "Média", "Baixa"]

df["Classificação"] = np.select(condicoes, resultados, default="Desconhecido")

# df.loc[df["Qualidade"] == 5].head()
df.head()
```

```python
# 5. Separar os dados de alvo (classificação)
from sklearn.preprocessing import StandardScaler

X = df[df.columns.difference(['Classificação', 'Qualidade'])].values
print(f'X: {X}')

scaler = StandardScaler()
X = scaler.fit_transform(X)

y = df['Classificação'].values
print(f'Y: {y}')
```

```python
# 6. Treinar o modelo KNN (escolha aleatoriamente o numero de vizinhos)
from sklearn.neighbors import KNeighborsClassifier

vinho_classificador = KNeighborsClassifier(n_neighbors=5)

vinho_classificador.fit(X, y)
```

```python
# 7. Confira a acurácia do algoritmo
from sklearn.model_selection import cross_val_score

scores_dt = cross_val_score(vinho_classificador, X, y, scoring='accuracy', cv=5)

print(scores_dt.mean())
```

```python
# 8. Teste com outros valores de vizinhos buscando melhorar a acurácia.

maior_acuracia_i = 0
maior_acuracia_mean = 0
for i in range(1, 100):
  vinho_classificador = KNeighborsClassifier(n_neighbors= i)
  vinho_classificador.fit(X, y)
  scores_dt = cross_val_score(vinho_classificador, X, y, scoring='accuracy', cv=5)

  if scores_dt.mean() > maior_acuracia_mean:
    maior_acuracia_i = i
    maior_acuracia_mean = scores_dt.mean()

  print(f'VIzinhos: {i} Acurácia: {scores_dt.mean()}')


print(f"Maior acurácia: {maior_acuracia_mean}, usando {maior_acuracia_i} vizinhos")
```
