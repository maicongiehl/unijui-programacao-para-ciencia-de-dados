## 8.4 - Compare os resultados e indique qual indicador apresenta maior variação relativa ao longo da semana.

```python
variacao_relativa_indicadores = np.diff(dados, axis=0) / dados[:-1]
oscilacao_por_indicador = np.max(np.abs(variacao_relativa_indicadores), axis=0)
indicador_vencedor = np.argmax(oscilacao_por_indicador)

mensagem = ""
if indicador_vencedor == 0:
  mensagem = "Quantidade de pedidos"
elif indicador_vencedor == 1:
  mensagem = "Faturamento (R$)"
elif indicador_vencedor == 2:
  mensagem = "Número de clientes"
elif indicador_vencedor == 3:
  mensagem = "Valor médio dos pedidos"
elif indicador_vencedor == 4:
  mensagem = "Quantidade de reclamações"

print(f"O indicador com maior variação foi: {mensagem}")
print(f"Oscilação: {oscilacao_por_indicador[indicador_vencedor]}")
```

## 9.1 - Qual foi o melhor dia da semana considerando o faturamento?

```python
index_melhor_dia_faturamento = np.argmax(dados[:, 1])
dia_maior_faturamento = dias[index_melhor_dia_faturamento]

print(f"O melhor dia da semana considerando o faturamento foi: {dia_maior_faturamento}")
```

## 9.2 - Esse mesmo dia também apresentou a maior quantidade de pedidos?

```python
index_melhor_dia_semana_qtd_pedidos = np.argmax(dados[:, 0])
dia_maior_qtd_pedidos = dias[index_melhor_dia_semana_qtd_pedidos]

if dia_maior_faturamento == dia_maior_qtd_pedidos:
  print(f"O dia que apresentou o maior número de pedidos converge com o dia de maior faturamento")
else:
  print(f"O dia que apresentou o maior número de pedidos diverge do dia de maior faturamento")
  print(f"Dia com a maior quantidade de pedidos: {dia_maior_qtd_pedidos}")
```

## 9.3 - O dia com maior faturamento foi também o dia com maior número de reclamações?

```python
index_dia_maior_qtd_reclamacoes = np.argmax(dados[:, 4])
dia_maior_qtd_reclamacoes = dias[index_dia_maior_qtd_reclamacoes]

if dia_maior_faturamento == dia_maior_qtd_reclamacoes:
  print(f"O dia que apresentou o maior número de reclamações converge com o dia de maior faturamento")
else:
  print(f"O dia que apresentou o maior número de reclamações diverge do dia de maior faturamento")
  print(f"Dia com a maior quantidade de pedidos: {dia_maior_qtd_reclamacoes}")
```
