# Escrever um programa usando Selenium para Scrap da temperatura (mínima e máxima) do dia, de uma cidade qualquer (informada pelo usuário), de algum site de previsão do tempo.

```python
import google_colab_selenium as gs
from selenium.webdriver.common.by import By
from time import sleep
```

```python
print("=" * 50)
uf = input("Digite a sigla da UF: ")
mun = input("Digite o nome do munícipio: ")
print("=" * 50)
```

```python
url = f"https://hgbrasil.com/weather/places/brazil/{uf}/{mun}"
navegador = gs.Chrome()
```

```python
navegador.get(url)
sleep(2)
```

```python
import re

def remove_non_numeric_from_str(value: str):
  return re.sub("[^0-9]", "", value)
```

```python
#  Busca máxima
max = remove_non_numeric_from_str(
    navegador.find_element(
      By.XPATH,
      '//*[@id="main"]/article/div/div/section[1]/div[4]/div[2]/span[1]'
).text)
```

```python
#  Busca mínima
min = remove_non_numeric_from_str(
  navegador.find_element(
    By.XPATH,
    '//*[@id="main"]/article/div/div/section[1]/div[4]/div[2]/span[2]'
  ).text
)
```

```python
print("=" * 50)
print(f'Temperatura em {mun.upper()} ({uf.upper()}):')
print("-" * 50)
print(f'Máxima: {max}°C')
print(f'Mínima: {min}°C')
print("=" * 50)
```
