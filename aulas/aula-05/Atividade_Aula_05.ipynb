# Atividade I
Escreva um programa que receba do usuário os valores de um conjunto de características de um personagem inventado por você.
Leia os valores e grave um arquivo para posterior manipulação.
Escolha o formato de gravação.

O arquivo não precisa ser alterado, apenas sobreescrito.
Crie uma função para ler e apresentar (print) os dados gravados.

```python
# Receber dados
nome = input("Digite o nome do personagem: ")
classe = input("Digite a classe do personagem: ")
pontos_vida = input("Digite a quantidade de pontos de vida do personagem: ")
```

```python
# Escrever arquivo
import json

with open('./personagem.json', 'w') as file:
  json.dump({'nome': nome, 'classe': classe, 'pontos_vida': pontos_vida}, file)
```

```python
# Função para ler arquivo
def ler_arquivo_personagem():
  with open("./personagem.json", "r") as file:
    personagem = json.load(file)
    return personagem
```

```python
# Print dos dados
personagem = ler_arquivo_personagem()
print(personagem)
```

# Atividade II
Gere uma lista com 20 números aleatórios e, para cada valor, calcule o resultado da multiplicação por 2 e por 10. Salve os dados em TXT, JSON, e CSV contendo três campos: valor original, valor *2, valor *10.

Em ambas as atividades
Ao final , adicione os arquivos gerados ao repositório do GitHub e realize o comit e push das alterações.

```python
# Gera lista
import random

lista = []

for i in range(20):
  numero = random.randrange(1, 100)

  lista.append(
      {
          "numero": numero,
          "x_2": numero * 2,
          "x_10": numero * 10,
      })
```

```python
# Salva em TXT
with open('./dados.txt', 'w') as file:
  for i in lista:
    numero = i.get('numero')
    file.write(f'N = {numero}, {numero} * 2 = {i.get('x_2')}, {numero} * 10 = {i.get('x_10')}\n')
```

```python
# Salva em JSON
import json

with open('./dados.json', 'w') as file:
  json.dump(lista, file)
```

```python
# Salva em CSV
import csv

nome_arquivo = "./dados.csv"
colunas = ["numero", "x_2", "x_10"]

with open(nome_arquivo, mode="w", newline="") as arquivo:
  writer = csv.DictWriter(arquivo, fieldnames=colunas)
  writer.writeheader()
  writer.writerows(lista)
```
