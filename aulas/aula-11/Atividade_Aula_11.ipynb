```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dados = {
    "Mes": ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun"],
    "Notebook": [20, 25, 22, 30, 35, 40],
    "Mouse": [50, 55, 48, 60, 65, 70],
    "Teclado": [30, 28, 35, 40, 38, 45]
}

df = pd.DataFrame(dados)
print(df)
```

# Questões

## 1. Exploração: Exiba as primeiras linhas, as dimensões do DataFrame e as estatísticas descritivas.

```python
print(f"Primeiras linhas:\n{df.head()}")
print(f"\nNúmero de dimensões: {df.ndim}")
print(f"\nEstatistícas:\n{df.describe()}")
```

## 2. Indicadores: Calcule com NumPy a média, o maior valor e o menor valor de vendas de cada produto.

```python
print("========== Indicadores ==========")
print("---------------------------------")
print("Produto: Notebook")
print(f"Média: {df['Notebook'].mean()}")
print(f"Máximo: {df['Notebook'].max()}")
print(f"Mínimo: {df['Notebook'].min()}")
print("---------------------------------")
print("Produto: Mouse")
print(f"Média: {df['Mouse'].mean()}")
print(f"Máximo: {df['Mouse'].max()}")
print(f"Mínimo: {df['Mouse'].min()}")
print("---------------------------------")
print("Produto: Teclado")
print(f"Média: {df['Teclado'].mean()}")
print(f"Máximo: {df['Teclado'].max()}")
print(f"Mínimo: {df['Teclado'].min()}")
print("---------------------------------")
print("=================================")
```

## 3. Evolução temporal: Crie um gráfico de linhas comparando as vendas dos três produtos ao longo dos meses.

```python
plt.plot(df["Mes"], df["Notebook"], label="Notebook")
plt.plot(df["Mes"], df["Mouse"], label="Mouse")
plt.plot(df["Mes"], df["Teclado"], label="Teclado")

plt.title("Evolução Mensal de Vendas por Produto")
plt.xlabel("Mês")
plt.ylabel("Qtd Vendas")

plt.legend(title="Produtos")

plt.show()
```

## 4. Comparação: Crie um gráfico de barras com o total vendido por produto no período.

```python
totais = df[["Notebook", "Mouse", "Teclado"]].sum()

barras = plt.bar(totais.index, totais.values, color=["dodgerblue", "darkorange", "green"])

plt.title("Total de Vendas por Produto (1º Semestre)")
plt.xlabel("Produto")
plt.ylabel("Total Vendido")

plt.show()
```

## 5. Distribuição: Construa um histograma com os valores mensais de vendas de um produto escolhido.

```python
produto = "Notebook"

plt.hist(df[produto], bins=4, color="darkorange", edgecolor="black")

plt.title(f"Distribuição das Vendas Mensais - {produto}")
plt.xlabel("Faixa de Quantidade Vendida")
plt.ylabel("Frequência (Meses)")

plt.show()
```

## 6. Interpretação
Identifique o produto com maior volume total, o mês de maior venda de cada produto e uma conclusão baseada nos gráficos.

```python
produtos = ["Notebook", "Mouse", "Teclado"]

# 1. Produto com maior volume total
totais = df[produtos].sum()
produto_lider = totais.idxmax()
maior_total = totais.max()

print(f"Produto com maior volume total: {produto_lider} ({maior_total} unidades)\n")

# 2. Mês de maior venda de cada produto
print("Mês de maior venda por produto:")
for produto in produtos:
    idx_max = df[produto].idxmax()
    mes_max = df.loc[idx_max, "Mes"]
    valor_max = df.loc[idx_max, produto]
    print(f"* {produto}: {mes_max} ({valor_max} unidades)")


print(f"\nConclusão")
print("Os três produtos cresceram ao longo do primeiro semestre, atingindo o pico ao fim do período.\n"+
      "Mouse lidera o volume de vendas em todos os meses, enquanto o Notebook apresenta o menor volume absoluto,\n"+
      "mas o maior crescimento proporcional (crescendo de 20 para 40 unidades vendidas)")
```

### Pergunta explorada
Como a participação percentual de cada produto no volume total de vendas mudou ao longo do semestre?

### Explicação
Utilizamos o gráfico de barras empilhadas para que todos os meses fiquem na mesma escala proporcional, permitindo enxergar a mudança da composição das vendas. Caso a informação de valor estivesse disponível, seria mais relevante, pois conseguiríamos enxergar a composição do resultado líquido, onde 1 venda a mais de notebook representaria uma porcentagem da composição muito maior.

Observa-se que o notebook aumentou levemente sua participação, ganhando espaço sobre o mouse. Teclado manteve-se estável.

```python
produtos = ["Notebook", "Mouse", "Teclado"]
df_pct = df[produtos].div(df[produtos].sum(axis=1), axis=0) * 100

base = np.zeros(len(df))

for produto in produtos:
    plt.bar(df["Mes"], df_pct[produto], bottom=base, label=produto)
    base += df_pct[produto]

plt.title("Participação Mensal nas Vendas por Produto (%)")
plt.ylabel("Participação (%)")
plt.legend(loc="upper right")
plt.show()
```
