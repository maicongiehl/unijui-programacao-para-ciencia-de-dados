```python
class Calculadora():
  def __init__(self):
    return None

  def somar(self, x, y):
    return x + y

  def subtrair(self, x, y):
    return x - y

  def multiplicar(self, x, y):
    resultado = 0
    for i in range(y):
      resultado += x
    return resultado

  def dividir(self, dividendo, divisor):
    if divisor == 0:
        raise ValueError("Não é possível dividir por zero.")

    # Sinal do resultado final
    negativo = (dividendo < 0) != (divisor < 0)

    # Garante que sejam valores positivos
    dividendo = abs(dividendo)
    divisor = abs(divisor)

    quociente = 0

    while dividendo >= divisor:
        dividendo = self.subtrair(dividendo, divisor)
        quociente = self.somar(quociente, 1)

    if negativo:
        quociente = -quociente

    return quociente

  def exponenciacao(self, base, expoente):
    # Qualquer número elevado a 0 é 1
    if expoente == 0:
        return 1

    expoente_negativo = expoente < 0
    expoente = abs(expoente)

    resultado = 1
    contador = 0

    while contador < expoente:
        resultado = self.multiplicar(resultado, base)
        contador = self.somar(contador, 1)

    # Se expoente negativo, 1/resultado
    if expoente_negativo:
        return self.dividir(1, resultado)

    return resultado

  def p(self, b, n):
      div = self.dividir(n, b)
      soma = self.somar(b, div)
      return self.dividir(soma, 2)

  def raiz(self, n):
      if n < 0:
          raise ValueError("Não existe raiz real de número negativo.")
      if n == 0:
          return 0

      b = 1
      while True:
          b = self.p(b, n)
          quadrado = self.exponenciacao(b, 2)
          diferenca = self.subtrair(quadrado, n)

          if abs(diferenca) < 0.0001:
              break

      return b
```

```python
calculadora = Calculadora()

while True:
  print("\n=======================")
  operacao = int(input(
    "[1] Soma\n" +
    "[2] Subtrair\n" +
    "[3] Multiplicar\n" +
    "[4] Dividir\n" +
    "[5] Exponenciação\n" +
    "[6] Raiz\n" +
    "[0] Sair\n" +
    "-----------------------\n" +
    "Digite a opção desejada: "
  ))

  resultado = 0
  if operacao == 0:
    break
  elif operacao == 1:
    # Soma
    x = int(input("Digite o primeiro número: "))
    y = int(input("Digite o segundo número: "))
    resultado = calculadora.somar(x, y)
  elif operacao == 2:
    # Subtração
    x = int(input("Digite o primeiro número: "))
    y = int(input("Digite o segundo número: "))
    resultado = calculadora.subtrair(x, y)
  elif operacao == 3:
    # Multiplicação
    x = int(input("Digite o primeiro número: "))
    y = int(input("Digite o segundo número: "))
    resultado = calculadora.multiplicar(x, y)
  elif operacao == 4:
    # Divisão
    x = int(input("Digite o dividendo: "))
    y = int(input("Digite o divisor: "))
    resultado = calculadora.dividir(x, y)
  elif operacao == 5:
    # Exponenciação
    x = int(input("Digite a base: "))
    y = int(input("Digite o expoente: "))
    resultado = calculadora.exponenciacao(x, y)
  elif operacao == 6:
    # Raíz
    x = int(input("Digite o número: "))
    resultado = calculadora.raiz(x)

  print("Resultado: " + str(resultado))
  print("-----------------------")
  print("=======================")
```
