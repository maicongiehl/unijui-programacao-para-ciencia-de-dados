# Desafio 1 – Análise de notas de uma turma
Desenvolva um programa que permita registrar as notas de uma turma de estudantes.

O programa deverá:
* solicitar a quantidade de estudantes;
* armazenar as notas em uma lista;

calcular:
* maior nota;
* menor nota;
* média da turma;
* quantidade de estudantes aprovados (nota ≥ 6,0);
* quantidade de estudantes reprovados.

Os cálculos deverão ser implementados utilizando funções.

```python
def maior_nota(notas: list) -> float:
  print("Procurando a maior nota...")
  resultado = max(notas)
  return resultado
```

```python
def menor_nota(notas: list) -> float:
  print("Procurando a menor nota...")
  resultado = min(notas)
  return resultado
```

```python
def media(notas: list) -> float:
  print("Calculando a média da turma...")
  resultado = sum(notas) / len(notas)
  return resultado
```

```python
def qtd_estudantes_aprovados(notas: list) -> int:
  print("Contando a quantidade de estudantes aprovados (nota >= 6.0)...")
  resultado = 0
  for nota in notas:
    if nota >= 6.0:
      resultado += 1

  return resultado
```

```python
def qtd_estudantes_reprovados(notas: list) -> int:
  print("Contando a quantidade de estudantes reprovados (nota < 6.0)...")
  resultado = 0
  for nota in notas:
    if nota < 6.0:
      resultado += 1

  return resultado
```

```python
print("=====================================")
qtd_estudantes = int(input("Digite a quantidade de estudantes: "))
print("-------------------------------------")

notas = []
for i in range(qtd_estudantes):
  nota = float(input(f'Digite a nota do estudante #{i+1}: '))
  notas.append(nota)

print("=====================================")


print("\n=====================================")
print("Executando cálculos...")
print("-------------------------------------")
maior_nota_resultado = maior_nota(notas)
menor_nota_resultado = menor_nota(notas)
media_resultado = media(notas)
qtd_estudantes_aprovados_resultado = qtd_estudantes_aprovados(notas)
qtd_estudantes_reprovados_resultado = qtd_estudantes_reprovados(notas)
print("=====================================")

print("\n=====================================")
print("Cálculos executados com sucesso!")
print("-------------------------------------")
print(f'Maior nota: {maior_nota_resultado}')
print(f'Menor nota: {menor_nota_resultado}')
print(f'Média da turma: {media_resultado}')
print(f'Quantidade de estudantes aprovados: {qtd_estudantes_aprovados_resultado}')
print(f'Quantidade de estudantes reprovados: {qtd_estudantes_reprovados_resultado}')
print("=====================================")
```

# Desafio 3 – Sistema de Estatísticas
Desenvolva um programa que leia uma sequência de números informados pelo usuário até que seja digitado
um valor de encerramento.
Ao final, o programa deverá apresentar:
* quantidade de valores informados;
* média;
* mediana;
* maior valor;
* menor valor;
* quantidade de números pares;
* quantidade de números ímpares.

Organize todos os cálculos em funções independentes.

```python
def qtd_valores_informados(numeros: list) -> int:
  print("Contando a quantidade de valores informados...")
  resultado = len(numeros)
  return resultado
```

```python
def media(numeros: list) -> float:
  print("Calculando a média...")
  resultado = sum(numeros) / len(numeros)
  return resultado
```

```python
import statistics

def mediana(numeros: list) -> int:
  print("Calculando a mediana...")
  resultado = statistics.median(numeros)
  return resultado
```

```python
def maior_valor(numeros: list) -> int:
  print("Procurando o elemento de maior valor...")
  resultado = max(numeros)
  return resultado
```

```python
def menor_valor(numeros: list) -> int:
  print("Procurando o elemento de menor valor...")
  resultado = min(numeros)
  return resultado
```

```python
def qtd_valores_pares(numeros: list) -> int:
  print("Contando o número de valores pares...")
  resultado = 0
  for numero in numeros:
    if numero % 2 == 0:
      resultado += 1

  return resultado
```

```python
def qtd_valores_impares(numeros: list) -> int:
  print("Contando o número de valores impares...")
  resultado = 0
  for numero in numeros:
    if numero % 2 != 0:
      resultado += 1

  return resultado
```

```python
numeros = []

print("=====================================")
print("Escreva uma lista de números")
print("Para sair, digite 0")
print("-------------------------------------")

while True:
  numero = int(input("Digite um número: ") )

  if numero == 0:
    break

  numeros.append(numero)

print("=====================================")

print("\n=====================================")
print("Executando cálculos...")
print("-------------------------------------")
qtd_valores_informados_resultado = qtd_valores_informados(numeros)
media_resultado = media(numeros)
mediana_resultado = mediana(numeros)
maior_valor_resultado = maior_valor(numeros)
menor_valor_resultado = menor_valor(numeros)
qtd_valores_pares_resultado = qtd_valores_pares(numeros)
qtd_valores_impares_resultado = qtd_valores_impares(numeros)
print("=====================================")

print("\n=====================================")
print("Cálculos executados com sucesso!")
print("-------------------------------------")
print(f'Quantidade de valor informados: {qtd_valores_informados_resultado}')
print(f'Média: {media_resultado}')
print(f'Mediana: {mediana_resultado}')
print(f'Maior valor: {maior_valor_resultado}')
print(f'Menor valor: {menor_valor_resultado}')
print(f'Quantidade de valores pares: {qtd_valores_pares_resultado}')
print(f'Quantidade de valores ímpares: {qtd_valores_impares_resultado}')
print("=====================================")
```
