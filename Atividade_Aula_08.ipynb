# Usandos os métodos do Pandas, localize a sua cidade de residência e:
1. Mostre todas as informações
2. Qual o mês de temperatura mínima
3. Qual o mês de maior déficit hídrico
4. A média de temperatura pelas 4 estações do ano

```python
import pandas as pd

url = "https://raw.githubusercontent.com/andr3nf/clima_municipios_brasileiros/refs/heads/main/clima_historico_municipal.csv"
df = pd.read_csv(url)
```

```python
# 1. Todas as informações do munícipio (Santa Rosa, RS)
cidade_df = df[(df["nome_municipio"] == "Santa Rosa") & (df["uf"] == "RS")].iloc[0]

print("=== Informações do Município ===")
print(cidade_df)
```

```python
# 2. Mês de temperatura mínima
cols_tmin = [c for c in df.columns if c.startswith("tmin_")]
col_tmin_min = cidade_df[cols_tmin].astype(float).idxmin()
mes_tmin = col_tmin_min.replace("tmin_", "")
val_tmin = cidade_df[col_tmin_min]

print(f"\nMês de temperatura mínima: {mes_tmin.upper()} ({val_tmin} °C)")
```

```python
# 3. Mês de maior déficit hídrico
cols_def = [c for c in df.columns if c.startswith("def_")]
col_def_max = cidade_df[cols_def].astype(float).idxmax()
mes_def = col_def_max.replace("def_", "")
val_def = cidade_df[col_def_max]

print(f"Mês de maior déficit hídrico: {mes_def.upper()} ({val_def} mm/mês)")
```

```python
# 4. Média de temperatura pelas 4 estações do ano
estacoes = {
    "Verão": ["temp_media_dez", "temp_media_jan", "temp_media_fev"],
    "Outono": ["temp_media_mar", "temp_media_abr", "temp_media_mai"],
    "Inverno": ["temp_media_jun", "temp_media_jul", "temp_media_ago"],
    "Primavera": ["temp_media_set", "temp_media_out", "temp_media_nov"],
}

print("\n=== Média de Temperatura Média por Estação ===")
for estacao, cols in estacoes.items():
    media_estacao = cidade_df[cols].astype(float).mean()
    print(f"{estacao}: {media_estacao:.2f} °C")
```
